```python
from edu1sces.model import HeisenbergModel
from edu1sces.solver import solve, SolverParameters

# 2-site S=1/2 Heisenberg chain
N = 26
model = HeisenbergModel(
    spins={i: 0.5 for i in range(N)},
    exchange_xy={(i, (i + 1) % N): 1.0 for i in range(N)},
    exchange_z={(i, (i + 1) % N): 1.0 for i in range(N)},
)
params = SolverParameters(output_log=True, num_threads=6)
result = solve(model, total_sz=0.0, params=params)
print(f"Energy: {result.energy}")
print(f"Expected: -0.75")
print(f"Dim: {result.dim()}")
```

```python
Sz = model.make_local_op_sz(0)
```

```python
Sz.print()
```

```python
result.expectation_onsite(
    local_op=Sz * Sz,
    site=0,
)
```
